# 01 – Explore the YeaZ dataset
First look at the phase-contrast images and their manual segmentation masks.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from skimage import io
```

```python
data_dir = Path("..") / "data" / "raw" / "gold-standard-PhC-plus-2"
data_dir.exists()
```

```python
image_paths = sorted(data_dir.glob("*_im.tif"))
len(image_paths)
```

```python
image = io.imread(image_paths[5])
print(image_paths[5].name)
print("shape:", image.shape)
print("dtype:", image.dtype)
print("min:", image.min(), "max", image.max())
```

```python
plt.imshow(image, cmap="gray")
plt.colorbar()
plt.show()
```

```python
frame = image[0]
print("frame shape:", frame.shape)

plt.imshow(frame, cmap="gray")
plt.colorbar()
plt.show()
```
